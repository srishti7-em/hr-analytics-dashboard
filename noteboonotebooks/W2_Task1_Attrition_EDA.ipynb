```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Load cleaned dataset
df = pd.read_csv('C:/Users/LENOVO/hr-analytics-dashboard/hr_cleaned.csv')

print("Libraries imported and data loaded!")
print(f"Shape: {df.shape[0]} rows × {df.shape[1]} columns")
```

```python
# Chart 1 - Overall Attrition Rate (Pie Chart)
plt.figure(figsize=(6,6))
df['Attrition'].value_counts().plot(kind='pie', 
    autopct='%1.1f%%', 
    colors=['#2ecc71','#e74c3c'],
    labels=['Stayed','Left'],
    startangle=90)
plt.title('Overall Employee Attrition Rate', fontsize=14, fontweight='bold')
plt.ylabel('')
plt.tight_layout()
plt.savefig('chart1_attrition_pie.png', dpi=150)
plt.show()
print("Chart 1 saved!")
```

```python
# Chart 2 - Attrition by Department (Bar Chart)
plt.figure(figsize=(8,5))
dept_attr = df.groupby('Department')['AttritionEncoded'].mean() * 100
dept_attr.plot(kind='bar', color=['#e74c3c','#f39c12','#3498db'], edgecolor='black')
plt.title('Attrition Rate by Department (%)', fontsize=14, fontweight='bold')
plt.xlabel('Department')
plt.ylabel('Attrition Rate (%)')
plt.xticks(rotation=0)
for i, v in enumerate(dept_attr):
    plt.text(i, v + 0.3, f'{v:.1f}%', ha='center', fontweight='bold')
plt.tight_layout()
plt.savefig('chart2_attrition_dept.png', dpi=150)
plt.show()
print("Chart 2 saved!")
```

```python
# Chart 3 - Attrition by Job Role (Horizontal Bar)
plt.figure(figsize=(10,6))
role_attr = df.groupby('JobRole')['AttritionEncoded'].mean() * 100
role_attr.sort_values().plot(kind='barh', color='#e74c3c', edgecolor='black')
plt.title('Attrition Rate by Job Role (%)', fontsize=14, fontweight='bold')
plt.xlabel('Attrition Rate (%)')
plt.ylabel('Job Role')
plt.tight_layout()
plt.savefig('chart3_attrition_jobrole.png', dpi=150)
plt.show()
print("Chart 3 saved!")
```

```python
# Chart 4 - Attrition by Salary Band (Grouped Bar)
plt.figure(figsize=(8,5))
salary_attr = df.groupby('SalaryBand')['AttritionEncoded'].mean() * 100
salary_attr = salary_attr.reindex(['Low', 'Medium', 'High'])
salary_attr.plot(kind='bar', color=['#e74c3c','#f39c12','#2ecc71'], edgecolor='black')
plt.title('Attrition Rate by Salary Band (%)', fontsize=14, fontweight='bold')
plt.xlabel('Salary Band')
plt.ylabel('Attrition Rate (%)')
plt.xticks(rotation=0)
for i, v in enumerate(salary_attr):
    plt.text(i, v + 0.3, f'{v:.1f}%', ha='center', fontweight='bold')
plt.tight_layout()
plt.savefig('chart4_attrition_salary.png', dpi=150)
plt.show()
print("Chart 4 saved!")
```

```python
# Chart 5 - Attrition by Tenure Group (Bar Chart)
plt.figure(figsize=(8,5))
tenure_attr = df.groupby('TenureGroup')['AttritionEncoded'].mean() * 100
tenure_attr = tenure_attr.reindex(['0-2 yrs', '3-5 yrs', '6-10 yrs', '10+ yrs'])
tenure_attr.plot(kind='bar', color='#9b59b6', edgecolor='black')
plt.title('Attrition Rate by Tenure Group (%)', fontsize=14, fontweight='bold')
plt.xlabel('Tenure Group')
plt.ylabel('Attrition Rate (%)')
plt.xticks(rotation=0)
for i, v in enumerate(tenure_attr):
    plt.text(i, v + 0.3, f'{v:.1f}%', ha='center', fontweight='bold')
plt.tight_layout()
plt.savefig('chart5_attrition_tenure.png', dpi=150)
plt.show()
print("Chart 5 saved!")
```

```python
# Chart 6 - Employee Count by Department (Bar Chart)
plt.figure(figsize=(8,5))
dept_count = df['Department'].value_counts()
dept_count.plot(kind='bar', color=['#3498db','#2ecc71','#e74c3c'], edgecolor='black')
plt.title('Employee Count by Department', fontsize=14, fontweight='bold')
plt.xlabel('Department')
plt.ylabel('Number of Employees')
plt.xticks(rotation=0)
for i, v in enumerate(dept_count):
    plt.text(i, v + 5, str(v), ha='center', fontweight='bold')
plt.tight_layout()
plt.savefig('chart6_dept_count.png', dpi=150)
plt.show()
print("Chart 6 saved!")
```

```python
# Chart 7 - Department vs Average Monthly Income (Bar Chart)
plt.figure(figsize=(8,5))
dept_salary = df.groupby('Department')['MonthlyIncome'].mean()
dept_salary.plot(kind='bar', color=['#3498db','#2ecc71','#e74c3c'], edgecolor='black')
plt.title('Average Monthly Income by Department', fontsize=14, fontweight='bold')
plt.xlabel('Department')
plt.ylabel('Average Monthly Income ($)')
plt.xticks(rotation=0)
for i, v in enumerate(dept_salary):
    plt.text(i, v + 50, f'${v:,.0f}', ha='center', fontweight='bold')
plt.tight_layout()
plt.savefig('chart7_dept_salary.png', dpi=150)
plt.show()
print("Chart 7 saved!")
```

```python
# Chart 8 - Monthly Income Distribution (Histogram)
plt.figure(figsize=(8,5))
plt.hist(df['MonthlyIncome'], bins=30, color='#3498db', edgecolor='black')
plt.title('Monthly Income Distribution', fontsize=14, fontweight='bold')
plt.xlabel('Monthly Income ($)')
plt.ylabel('Number of Employees')
plt.tight_layout()
plt.savefig('chart8_income_histogram.png', dpi=150)
plt.show()
print("Chart 8 saved!")
```

```python
# Chart 9 - Salary Boxplot by Department
plt.figure(figsize=(8,5))
df.boxplot(column='MonthlyIncome', by='Department', 
           patch_artist=True,
           boxprops=dict(facecolor='#3498db', color='black'),
           medianprops=dict(color='red', linewidth=2))
plt.title('Salary Distribution by Department', fontsize=14, fontweight='bold')
plt.suptitle('')
plt.xlabel('Department')
plt.ylabel('Monthly Income ($)')
plt.tight_layout()
plt.savefig('chart9_salary_boxplot.png', dpi=150)
plt.show()
print("Chart 9 saved!")
```

```python
# Chart 10 - Salary Boxplot by Job Role
plt.figure(figsize=(12,6))
df.boxplot(column='MonthlyIncome', by='JobRole',
           patch_artist=True,
           boxprops=dict(facecolor='#2ecc71', color='black'),
           medianprops=dict(color='red', linewidth=2))
plt.title('Salary Distribution by Job Role', fontsize=14, fontweight='bold')
plt.suptitle('')
plt.xlabel('Job Role')
plt.ylabel('Monthly Income ($)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('chart10_salary_jobrole_boxplot.png', dpi=150)
plt.show()
print("Chart 10 saved!")
```
