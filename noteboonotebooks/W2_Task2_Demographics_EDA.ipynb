```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Load cleaned dataset
df = pd.read_csv('C:/Users/LENOVO/hr-analytics-dashboard/hr_cleaned.csv')

print("Libraries imported and data loaded!")
print(f"Shape: {df.shape[0]} rows × {df.shape[1]} columns")
```

```python
# Chart 1 - Gender Distribution by Department (Stacked Bar)
plt.figure(figsize=(8,5))
gender_dept = df.groupby(['Department','Gender']).size().unstack()
gender_dept.plot(kind='bar', stacked=True, 
                 color=['#e74c3c','#3498db'], edgecolor='black')
plt.title('Gender Distribution by Department', fontsize=14, fontweight='bold')
plt.xlabel('Department')
plt.ylabel('Number of Employees')
plt.xticks(rotation=0)
plt.legend(title='Gender')
plt.tight_layout()
plt.savefig('chart11_gender_dept.png', dpi=150)
plt.show()
print("Chart 11 saved!")
```

```python
# Chart 2 - Age Distribution (Histogram)
plt.figure(figsize=(8,5))
plt.hist(df['Age'], bins=20, color='#9b59b6', edgecolor='black')
plt.title('Age Distribution of Employees', fontsize=14, fontweight='bold')
plt.xlabel('Age')
plt.ylabel('Number of Employees')
plt.tight_layout()
plt.savefig('chart12_age_distribution.png', dpi=150)
plt.show()
print("Chart 12 saved!")
```

```python
# Chart 3 - Age Group vs Attrition Rate (Bar)
plt.figure(figsize=(8,5))
age_attr = df.groupby('AgeGroup')['AttritionEncoded'].mean() * 100
age_attr.plot(kind='bar', color='#e74c3c', edgecolor='black')
plt.title('Attrition Rate by Age Group (%)', fontsize=14, fontweight='bold')
plt.xlabel('Age Group')
plt.ylabel('Attrition Rate (%)')
plt.xticks(rotation=0)
for i, v in enumerate(age_attr):
    plt.text(i, v + 0.3, f'{v:.1f}%', ha='center', fontweight='bold')
plt.tight_layout()
plt.savefig('chart13_age_attrition.png', dpi=150)
plt.show()
print("Chart 13 saved!")
```

```python
# Chart 4 - Gender vs Average Salary (DEI Analysis)
plt.figure(figsize=(6,5))
gender_salary = df.groupby('Gender')['MonthlyIncome'].mean()
gender_salary.plot(kind='bar', color=['#e74c3c','#3498db'], edgecolor='black')
plt.title('Average Salary by Gender (DEI Analysis)', fontsize=14, fontweight='bold')
plt.xlabel('Gender')
plt.ylabel('Average Monthly Income ($)')
plt.xticks(rotation=0)
for i, v in enumerate(gender_salary):
    plt.text(i, v + 50, f'${v:,.0f}', ha='center', fontweight='bold')
plt.tight_layout()
plt.savefig('chart14_gender_salary.png', dpi=150)
plt.show()
print("Chart 14 saved!")
```

```python
# Chart 5 - Performance Rating Distribution (Bar)
plt.figure(figsize=(6,5))
df['PerformanceRating'].value_counts().sort_index().plot(
    kind='bar', color='#2ecc71', edgecolor='black')
plt.title('Performance Rating Distribution', fontsize=14, fontweight='bold')
plt.xlabel('Performance Rating')
plt.ylabel('Number of Employees')
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig('chart15_performance_dist.png', dpi=150)
plt.show()
print("Chart 15 saved!")
```

```python
# Chart 6 - Performance vs Attrition (Grouped Bar)
plt.figure(figsize=(7,5))
perf_attr = df.groupby('PerformanceRating')['AttritionEncoded'].mean() * 100
perf_attr.plot(kind='bar', color='#f39c12', edgecolor='black')
plt.title('Attrition Rate by Performance Rating (%)', fontsize=14, fontweight='bold')
plt.xlabel('Performance Rating')
plt.ylabel('Attrition Rate (%)')
plt.xticks(rotation=0)
for i, v in enumerate(perf_attr):
    plt.text(i, v + 0.3, f'{v:.1f}%', ha='center', fontweight='bold')
plt.tight_layout()
plt.savefig('chart16_performance_attrition.png', dpi=150)
plt.show()
print("Chart 16 saved!")
```

```python
# Chart 7 - Work Life Balance vs Attrition (Scatter)
plt.figure(figsize=(8,5))
wlb_attr = df.groupby('WorkLifeBalance')['AttritionEncoded'].mean() * 100
wlb_attr.plot(kind='bar', color='#1abc9c', edgecolor='black')
plt.title('Attrition Rate by Work Life Balance Score', fontsize=14, fontweight='bold')
plt.xlabel('Work Life Balance (1=Low, 4=High)')
plt.ylabel('Attrition Rate (%)')
plt.xticks(rotation=0)
for i, v in enumerate(wlb_attr):
    plt.text(i, v + 0.3, f'{v:.1f}%', ha='center', fontweight='bold')
plt.tight_layout()
plt.savefig('chart17_worklife_attrition.png', dpi=150)
plt.show()
print("Chart 17 saved!")
```

```python
# Chart 8 - Years at Company Distribution (Histogram)
plt.figure(figsize=(8,5))
plt.hist(df['YearsAtCompany'], bins=20, color='#3498db', edgecolor='black')
plt.title('Years at Company Distribution', fontsize=14, fontweight='bold')
plt.xlabel('Years at Company')
plt.ylabel('Number of Employees')
plt.tight_layout()
plt.savefig('chart18_tenure_distribution.png', dpi=150)
plt.show()
print("Chart 18 saved!")
```

```python
# Chart 9 - Years Since Last Promotion vs Attrition (Scatter)
plt.figure(figsize=(8,5))
promo_attr = df.groupby('YearsSinceLastPromotion')['AttritionEncoded'].mean() * 100
plt.scatter(promo_attr.index, promo_attr.values, color='#e74c3c', s=100)
plt.plot(promo_attr.index, promo_attr.values, color='#e74c3c', alpha=0.5)
plt.title('Attrition Rate vs Years Since Last Promotion', fontsize=14, fontweight='bold')
plt.xlabel('Years Since Last Promotion')
plt.ylabel('Attrition Rate (%)')
plt.tight_layout()
plt.savefig('chart19_promotion_attrition.png', dpi=150)
plt.show()
print("Chart 19 saved!")
```

```python
# Chart 10 - Job Involvement vs Performance (Heatmap)
plt.figure(figsize=(8,5))
heatmap_data = df.groupby(['JobInvolvement','PerformanceRating']).size().unstack()
sns.heatmap(heatmap_data, annot=True, fmt='d', cmap='YlOrRd')
plt.title('Job Involvement vs Performance Rating (Heatmap)', fontsize=14, fontweight='bold')
plt.xlabel('Performance Rating')
plt.ylabel('Job Involvement')
plt.tight_layout()
plt.savefig('chart20_involvement_performance_heatmap.png', dpi=150)
plt.show()
print("Chart 20 saved!")
```

```python
# Chart 11 - Correlation Heatmap
plt.figure(figsize=(12,8))
numeric_cols = df.select_dtypes(include='number').columns
corr_matrix = df[numeric_cols].corr()
sns.heatmap(corr_matrix, annot=False, cmap='coolwarm', center=0)
plt.title('Correlation Heatmap of All Numeric Features', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('chart21_correlation_heatmap.png', dpi=150)
plt.show()
print("Chart 21 saved!")
```

```python
# Chart 12 - Age vs Monthly Income (Scatter coloured by Attrition)
plt.figure(figsize=(8,5))
colors = df['AttritionEncoded'].map({0:'#2ecc71', 1:'#e74c3c'})
plt.scatter(df['Age'], df['MonthlyIncome'], c=colors, alpha=0.5)
plt.title('Age vs Monthly Income (Green=Stayed, Red=Left)', fontsize=14, fontweight='bold')
plt.xlabel('Age')
plt.ylabel('Monthly Income ($)')
plt.tight_layout()
plt.savefig('chart22_age_income_scatter.png', dpi=150)
plt.show()
print("Chart 22 saved!")
```
