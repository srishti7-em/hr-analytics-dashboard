## 1. Executive Summary

This report presents key findings from the HR Analytics Dashboard built using the IBM HR Analytics Employee Attrition dataset (1,470 employees, 35 features). The analysis reveals critical workforce patterns related to attrition, compensation, diversity, and performance that can help HR leadership make data-driven decisions.

**3 Key Findings:**
- The overall attrition rate is **16.1%** — with Sales department showing the highest rate at **20.6%**
- Employees in the **Low salary band** are **2.5x more likely** to leave than High salary band employees
- Employees with **0-2 years tenure** have the highest attrition, indicating an onboarding and early engagement problem

## 2. Attrition Insights — Top 3 Risk Factors

### Risk Factor 1: Department
- Sales department has the highest attrition rate at **20.6%**
- Human Resources follows at **19.0%**
- Research & Development has the lowest at **13.8%**
- **Recommendation:** Sales team needs immediate retention intervention

### Risk Factor 2: Salary Band
- Low salary band employees have an attrition rate of **~26%**
- Medium salary band: **~17%**
- High salary band: **~7%**
- **Recommendation:** Review and revise compensation structure for low-band employees

### Risk Factor 3: Tenure
- Employees with 0-2 years tenure have the highest attrition
- Attrition drops significantly after 5+ years
- **Recommendation:** Implement a structured 90-day onboarding program and early engagement initiatives for new hires

## 3. Compensation Analysis

### Salary Distribution Findings:
- Average monthly income across all employees: **$6,502**
- Highest paid role: **Manager** (~$17,000/month)
- Lowest paid role: **Sales Representative** (~$2,600/month)
- Research Director and Manufacturing Director also among top earners

### Salary Equity Findings:
- **Gender pay gap observed:** Male employees earn slightly more on average than female employees
- Salary varies significantly by department — R&D employees earn more on average than Sales
- Employees in Low salary band are significantly more likely to leave

### Recommendations:
- Conduct a formal salary equity audit across gender and department
- Consider salary band adjustments for Sales Representatives who show highest attrition
- Link salary increments to performance ratings to improve retention

## 4. DEI Findings — Gender & Age Analysis

### Gender Distribution:
- Male employees: **60%** of workforce
- Female employees: **40%** of workforce
- Gender ratio is relatively balanced but male-dominated

### Age Distribution:
- Majority of employees fall in **26-35 age group** (largest segment)
- Average employee age: **36.92 years**
- Youngest age group (18-25) shows highest attrition rate
- Employees aged 46+ show lowest attrition — most loyal segment

### DEI Recommendations:
- Increase female representation especially in senior roles
- Create mentorship programs for younger employees (18-25) to improve retention
- Leverage experience of 46+ employees through knowledge transfer programs
- Set diversity hiring targets for next recruitment cycle

## 5. Performance Insights

### Key Findings:
- Majority of employees rated **3 (Excellent)** or **4 (Outstanding)**
- Very few employees rated 1 or 2 — suggesting rating inflation
- **Surprising finding:** High performers also leave — performance rating alone does not predict retention
- Employees with **low work-life balance score (1)** have significantly higher attrition (~31%)
- Job involvement strongly correlates with performance — highly involved employees perform better

### Performance vs Attrition:
- Attrition rate is similar across performance ratings
- This suggests employees leave for reasons other than performance — mainly compensation and work-life balance
- Tenure positively correlates with performance — longer-serving employees perform better

### Recommendations:
- Introduce work-life balance initiatives — flexible working, mental health support
- Re-calibrate performance rating system to reduce rating inflation
- Use job involvement scores to identify disengaged employees early
- Reward high performers with non-monetary benefits to improve retention

## 6. Five Actionable HR Recommendations

### Recommendation 1: Launch a Sales Retention Programme
**Problem:** Sales has 20.6% attrition — highest in the company
**Action:** Introduce performance bonuses, career development paths, and monthly 1-on-1 check-ins for Sales employees
**Expected Impact:** Reduce Sales attrition by 5-8% within 6 months

### Recommendation 2: Revise Compensation for Low Salary Band
**Problem:** Low salary band employees leave at 2.5x the rate of high earners
**Action:** Conduct salary benchmarking against industry standards and increase base pay for bottom 33% earners
**Expected Impact:** Reduce overall attrition rate from 16.1% to below 12%

### Recommendation 3: Improve Onboarding for New Hires (0-2 years)
**Problem:** Highest attrition occurs in first 2 years
**Action:** Implement structured 90-day onboarding, assign mentors, and conduct 30/60/90 day check-ins
**Expected Impact:** Improve first-year retention by 20%

### Recommendation 4: Introduce Work-Life Balance Initiatives
**Problem:** Employees with low work-life balance score leave more
**Action:** Introduce flexible working hours, remote work options, and mental health support programmes
**Expected Impact:** Improve work-life balance scores and reduce attrition among affected employees

### Recommendation 5: Address Gender Pay Gap
**Problem:** Gender pay gap observed in DEI analysis
**Action:** Conduct formal pay equity audit, set equal pay targets, and publish diversity metrics annually
**Expected Impact:** Improve gender equity, employer brand, and female retention
